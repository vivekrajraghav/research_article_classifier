```python
import pandas as pd
```

```python
df=pd.read_excel("Business/eco.xls")
```

```python
df_eco=df[["Abstract","Research Areas"]].copy()
df_eco.sample(10)
```

```python
exclusion_keywords = [
    'Science', 'Social Science', 'Sociology','Psychology', 'Anthropology',
    'Education & Educational Research', 'Behavioral','Food science',
    'Environmental Sciences & Ecology'," Women's Studies",'Government & Law','Mathematics','Family Studies']
pattern = '|'.join(exclusion_keywords)
df_eco = df_eco[~df_eco['Research Areas'].str.contains(pattern, case=False, na=False)].copy()
```

```python
df_eco["Research Areas"]="Business & Economics"
```

```python
df_eco["Research Areas"].value_counts()
```
