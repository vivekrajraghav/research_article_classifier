```python
import re
import nltk
from nltk.corpus import wordnet, stopwords
from nltk.stem import WordNetLemmatizer
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('averaged_perceptron_tagger')
```

```python
lemmatizer=WordNetLemmatizer()
stop_word=set(stopwords.words('english'))
additional_noise_words = [
    'social', 'science', 'business', 'economics',
    'paper', 'study', 'research', 'article', 'results', 'show', 'abstract', 
    'journal', 'author', 'conclusion', 'report', 'citation', 'evaluation', 
    'domain', 'concept', 'factor', 'public', 'scientific', 'scientist', 
    'data', 'information', 'publication', 'cite', 'analyze', 'come', 'people',
    'model', 'requirement', 'implementation', 'discovery', 'argue', 'scholar',
    'environment','association', 'publish', 'western', 'ib', 'library','sociology', 
    'sociologist', 'sociological', 'society'
]
stop_word.update(additional_noise_words)
def clean_text(text):
    text=str(text)
    text = re.sub(r"'s\b", "", text)
    text=text.replace("-"," ")
    text=re.sub(r'[^a-zA-Z\s]','',text)
    words=text.lower().split()
    tagged_word=nltk.pos_tag(words)
    cleaned_words=[]
    for word,tag in tagged_word:
        pos=tag[0].upper()
        tag_dict={
                "J":wordnet.ADJ,
                "N":wordnet.NOUN,
                "V":wordnet.VERB,
                "R":wordnet.ADV
            }
        wordnet_pos=tag_dict.get(pos,wordnet.NOUN)
        lemma_word=lemmatizer.lemmatize(word,wordnet_pos)
        if lemma_word not in stop_word:
            cleaned_words.append(lemma_word)
    return ' '.join(cleaned_words)
df["cleaned_abstract"]=df["Abstract"].apply(clean_text)
```

```python
print(f"Original abstract:\n{df["Abstract"].iloc[5]}\n")
print(f"Lemmatized abstract:\n{df["cleaned_abstract"].iloc[5]}\n")
```

```python
Y=df["Research Areas"]
X=df["cleaned_abstract"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=42,shuffle=True,stratify=Y)
```

```python
tfidf=TfidfVectorizer(ngram_range=(1,2),max_features=5000)
X_train_tfidf=tfidf.fit_transform(X_train)
X_test_tfidf=tfidf.transform(X_test)
```

```python
classifier={
    "Logistic Regression":LogisticRegression(max_iter=1000,random_state=42),
    "Decision Trees":DecisionTreeClassifier(random_state=42,max_depth=15),
    "Mutlinomial Naive Bayes":MultinomialNB(),
    "K-Nearest Neighbors(KNN)":KNeighborsClassifier(n_neighbors=15),
    "Suppport Vector Machine":LinearSVC(random_state=42),
    "Random Forest":RandomForestClassifier(n_estimators=100,random_state=42)}
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_dynamic_features(classifier, vectorizer, model_name, top_n=20):
    """Dynamically visualizes top features based on the algorithm's architecture."""
    feature_names = np.array(vectorizer.get_feature_names_out())
    
    # 1. Linear Models (SVM, Logistic Regression) - Class Specific
    if hasattr(classifier, 'coef_'):
        classes = classifier.classes_
        fig, axes = plt.subplots(1, len(classes), figsize=(15, 5), sharey=False)
        fig.suptitle(f'{model_name}: Top Words by Category (Coefficients)', fontsize=16)
        
        for i, class_label in enumerate(classes):
            coefs = classifier.coef_[i]
            top_indices = np.argsort(coefs)[-top_n:]
            axes[i].barh(feature_names[top_indices], coefs[top_indices], color='skyblue')
            axes[i].set_title(f'Top for {class_label}')
            axes[i].invert_yaxis()
            
        plt.tight_layout()
        plt.show()

    # 2. Naïve Bayes - Class Specific Log Probabilities
    elif hasattr(classifier, 'feature_log_prob_'):
        classes = classifier.classes_
        fig, axes = plt.subplots(1, len(classes), figsize=(15, 5), sharey=False)
        fig.suptitle(f'{model_name}: Top Words by Category (Log Probabilities)', fontsize=16)
        
        for i, class_label in enumerate(classes):
            log_prob = classifier.feature_log_prob_[i]
            top_indices = np.argsort(log_prob)[-top_n:]
            axes[i].barh(feature_names[top_indices], log_prob[top_indices], color='lightgreen')
            axes[i].set_title(f'Top for {class_label}')
            axes[i].invert_yaxis()
            
        plt.tight_layout()
        plt.show()

    # 3. Tree Models (Random Forest, Decision Tree) - Global Importance
    elif hasattr(classifier, 'feature_importances_'):
        fig, ax = plt.subplots(figsize=(8, 5))
        importances = classifier.feature_importances_
        top_indices = np.argsort(importances)[-top_n:]
        
        ax.barh(feature_names[top_indices], importances[top_indices], color='salmon')
        ax.set_title(f'{model_name}: Top Words (Global Feature Importance)', fontsize=16)
        ax.invert_yaxis()
        
        plt.tight_layout()
        plt.show()

    # 4. kNN - No inherent feature importance
    else:
        print(f"Skipping filter visualization: '{model_name}' does not extract feature weights.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd
from sklearn.metrics import confusion_matrix
for name, clf in classifier.items():
    print(f"--------------------{name}------------------------")
    clf.fit(X_train_tfidf,Y_train)
    Y_pred=clf.predict(X_test_tfidf)
    print(f"Accuracy: {100*accuracy_score(Y_test,Y_pred):.3f}%") 
    print(classification_report(Y_test,Y_pred))
    cm = confusion_matrix(Y_test, Y_pred)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Reds',
                xticklabels=clf.classes_, 
                yticklabels=clf.classes_)
    plt.title(f'{name}')
    plt.ylabel('Actual Category')
    plt.xlabel('Predicted Category')
    plt.show()
    plot_dynamic_features(clf,tfidf,name)
    print("-"*100+"\n")
```

```python
from sklearn.pipeline import Pipeline
import pickle
from sklearn.preprocessing import FunctionTransformer
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import learning_curve

train_sizes, train_scores, test_scores = learning_curve(
    estimator=pipeline,     
    X=X_train, 
    y=Y_train,
    cv=5,                      
    n_jobs=-1,               
    train_sizes=np.linspace(0.1, 1.0, 10),
    scoring='accuracy'
)

train_mean = np.mean(train_scores, axis=1)
train_std = np.std(train_scores, axis=1)
test_mean = np.mean(test_scores, axis=1)
test_std = np.std(test_scores, axis=1)

plt.figure(figsize=(10, 6))


plt.plot(train_sizes, train_mean, color='blue', marker='o', label='Training Accuracy')
plt.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.15, color='blue')


plt.plot(train_sizes, test_mean, color='green', marker='s', linestyle='--', label='Cross-Validation Accuracy')
plt.fill_between(train_sizes, test_mean - test_std, test_mean + test_std, alpha=0.15, color='green')

plt.title('Learning Curve with 5-Fold Cross-Validation', fontsize=14, fontweight='bold')
plt.xlabel('Number of Training Abstracts', fontsize=12)
plt.ylabel('Accuracy Score', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(loc='lower right', fontsize=12)
plt.tight_layout()
plt.show()
```
